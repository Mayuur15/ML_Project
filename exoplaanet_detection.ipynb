```python
#raw data
plt.figure(figsize=(12, 5))

lc_collection.plot();

plt.title("Raw Observations of Kepler-69")
plt.show()
```

```python
#after stitching and removing nans2
raw_lc = lc_collection.stitch()
raw_lc = raw_lc.remove_nans()
plt.figure(figsize=(12, 5))
raw_lc.plot()

plt.title("Combined Light Curve of Kepler-69")
plt.xlabel("Time (Days)")
plt.ylabel("Flux")
plt.show()
```

```python
#removing outliers
clean_lc = raw_lc.remove_outliers(sigma=5)

print("Original stuff", len(raw_lc.time))
print("After i remove outliers:", len(clean_lc.time))
print( len(raw_lc.time) - len(clean_lc.time))
```

```python
# after flattening 
flat_lc, trend_lc = clean_lc.flatten(window_length=901,return_trend=True)
```

```python
#comparing 

fig, axes = plt.subplots(2, 1, figsize=(14, 9))
clean_lc.plot(ax=axes[0])
axes[0].set_title("Before Detrending")
axes[0].set_ylabel("Flux")
flat_lc.plot(ax=axes[1])
axes[1].set_title("After Detrending")
axes[1].set_ylabel("Normalized Flux")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 5))

trend_lc.plot()

plt.title("Estimated Stellar Brightness Trend")
plt.ylabel("Flux")
plt.show()
```

```python
bls_lc = flat_lc.remove_nans()
```

```python
periods = np.exp(np.linspace(np.log(0.61), np.log(300), 100000))
durations = np.array([0.05, 0.1, 0.15, 0.2, 0.25, 0.3, 0.4, 0.5, 0.52, 0.55, 0.6])
```

```python
bls = bls_lc.to_periodogram(method='bls', period=periods, duration=durations)

bls.plot()
plt.title("BLS Periodogram - Kepler-69")
plt.xlabel("Orbital Period (Days)")
plt.ylabel("BLS Power")
plt.show()
```

```python
best_period = bls.period_at_max_power
t0 = bls.transit_time_at_max_power
dur = bls.duration_at_max_power
print("Period:", best_period)
print("T0:", t0)
print("Duration:", dur)
print("Depth:", bls.depth_at_max_power)
print("Max power:", bls.max_power)
```

```python
folded = bls_lc.fold(period=best_period, epoch_time=t0)

fig, ax = plt.subplots(1, 2, figsize=(14, 5))
folded.scatter(ax=ax[0], s=2, alpha=0.3)
ax[0].set_title(f"Folded at P = {best_period:.4f}")
folded.bin(time_bin_size=0.02).scatter(ax=ax[1], s=8)
ax[1].set_xlim(-0.5, 0.5)
ax[1].set_title("Zoom on transit (binned)")
plt.show()
```

```python
mask = bls.get_transit_mask(period=best_period, transit_time=t0, duration=dur * 1.5)
masked_lc = bls_lc[~mask]

# Search for the second signal
bls2 = masked_lc.to_periodogram(method='bls', period=periods, duration=durations)
bls2.plot()
plt.title("BLS after masking first signal")
plt.xlabel("Orbital Period (Days)")
plt.ylabel("BLS Power")
plt.show()

print("Second period:", bls2.period_at_max_power)
print("Depth:", bls2.depth_at_max_power)
```

```python
from sklearn.ensemble import IsolationForest
import numpy as np
import matplotlib.pyplot as plt
flux = flat_lc.flux.value
model = IsolationForest(
    contamination=0.02,
    random_state=42
)
predictions = model.fit_predict(
    flux.reshape(-1, 1)
)
anomalies = predictions == -1
plt.figure(figsize=(12, 5))

plt.plot(flat_lc.time.value, flux,
         label="Light curve", linewidth=0.5)

plt.scatter(
    flat_lc.time.value[anomalies],
    flux[anomalies],
    color="red",
    s=8,
    label="ML detected anomalies"
)

plt.xlabel("Time (days)")
plt.ylabel("Normalized Flux")
plt.title("Isolation Forest Anomaly Detection")
plt.legend()
plt.show()
```

```python
best_period = bls.period_at_max_power

print("Best orbital period:", best_period)
print("Maximum BLS power:", bls.max_power)
```

```python
# Select only unusually LOW brightness measurements
low_anomalies = anomalies & (flux < np.median(flux))

# Create a light curve containing those measurements
anomaly_lc = flat_lc[low_anomalies]

# Fold both light curves using the BLS period
folded_lc = flat_lc.fold(
    period=best_period,
    epoch_time=bls.transit_time_at_max_power
)

folded_anomaly_lc = anomaly_lc.fold(
    period=best_period,
    epoch_time=bls.transit_time_at_max_power
)

# Plot
plt.figure(figsize=(10, 5))

plt.scatter(
    folded_lc.time.value,
    folded_lc.flux.value,
    s=2,
    label="All observations"
)

plt.scatter(
    folded_anomaly_lc.time.value,
    folded_anomaly_lc.flux.value,
    color="red",
    s=12,
    label="Low-flux ML anomalies"
)

plt.xlabel("Orbital phase (days)")
plt.ylabel("Normalized Flux")
plt.title("ML Anomalies Compared with BLS Transit Period")
plt.legend()
plt.show()
```

```python
import astropy.units as u
import numpy as np

# Get orbital phase of all observations
phase_all = folded_lc.time.value

# Get orbital phase of low-flux anomalies
phase_anomaly = folded_anomaly_lc.time.value

# Get transit duration from BLS
duration = bls.duration_at_max_power.to_value(u.day)

# Define the predicted transit window
transit_window = duration / 2

# Count observations inside and outside transit window
inside_all = np.abs(phase_all) <= transit_window
inside_anomaly = np.abs(phase_anomaly) <= transit_window

# Calculate anomaly rates
inside_rate = np.sum(inside_anomaly) / np.sum(inside_all) * 100

outside_rate = (
    (len(phase_anomaly) - np.sum(inside_anomaly)) /
    (len(phase_all) - np.sum(inside_all))
) * 100

print("BLS transit duration:", duration, "days")
print("Anomaly rate inside transit:", inside_rate, "%")
print("Anomaly rate outside transit:", outside_rate, "%")
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    folded_lc.time.value,
    folded_lc.flux.value,
    s=2,
    alpha=0.3,
    label="All observations"
)

plt.scatter(
    folded_anomaly_lc.time.value,
    folded_anomaly_lc.flux.value,
    color="red",
    s=10,
    label="ML anomalies"
)

duration = bls.duration_at_max_power.to_value(u.day)

plt.axvspan(
    -duration/2,
    duration/2,
    alpha=0.2,
    color="green",
    label="BLS transit window"
)

plt.xlim(-0.5, 0.5)

plt.xlabel("Orbital phase (days)")
plt.ylabel("Normalized Flux")
plt.title("Zoomed Transit Region: BLS and ML Comparison")
plt.legend()
plt.show()
```
